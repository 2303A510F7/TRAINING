```python
#implementation of depth first traversal-DFT
def DFT(visited,graph,node):
    if node not in visited:
        print(node,end=" ")
        visited.add(node)
        for adjacentnodes in graph[node]:
          if adjacentnodes not in visited:
            DFT(visited,graph,adjacentnodes)
visited=set()
graph={
    '5':['3','7'],
    '3':['2','4'],
    '4':['8'],
    '7':['8'],
    '2':[],
    '8':[]
}
DFT(visited,graph,'5')
```

```python
from collections import deque
Q=deque()
# implementation of breadth first traversal - BFT
#Function definition
def BFT(visited, graph, node):
    visited.add(node)
    Q.appendleft(node)
    while Q:
      node=Q.pop()
      print(node,end=" ")
      for adjacentnodes in graph[node]:
        if adjacentnodes not in visited:
          visited.add(adjacentnodes)
          Q.appendleft(adjacentnodes)
#driver code
visited = set()
graph = {
    'A':['B','C'],
    'B':['D','E'],
    'C':['F','G'],
    'D':[],
    'E':[],
    'F':[],
    'G':[]
}
BFT(visited, graph, 'A')
```
