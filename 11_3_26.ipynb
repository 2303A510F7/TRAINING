QUEUE USING LINKED LIST

```python
class Node:
  def __init__(self,data):
    self.data=data
    self.next=None
  #queue using linked list
class Queue:
  def __init__(self):
    self.head=None
    #insert at end (enqueue)
  def enqueue(self,data):
    new_node=Node(data)
    if self.head is None:
      self.head=new_node
      return
    temp=self.head
    while temp.next:
      temp=temp.next
    temp.next=new_node
    #delete
  def dequeue(self):
    if self.head is None:
      print("Queue is empty")
      return
      temp=self.head
      self.head=temp.next
      temp=None

```

stack implementaion using queue(first in first out)

```python
class QueueOnestack:
  def __init__(self):
    self.stack=[]
  def enqueue(self,data):
    self.stack.append(data)
  def dequeue(self):
    if not self.stack:
      return "Queue is empty"
    if len(self.stack)==1:
      return self.stack.pop()#here last value is stored in item
    temp=self.stack.pop()
    item=self.dequeue()
    self.stack.append(temp)
    return item
  def isEmpty(self):
    return len(self.stack)==0
q=QueueOnestack()
q.dequeue()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
print(q.dequeue())
print(q.dequeue())
print(q.dequeue())
print(q.dequeue())
print(q.dequeue())
```

```python
#reverse a queue -using stack
from collections import deque
def reverse_queue(q):
  stack=[]
  #move queue elements to stack
  while q:
    stack.append(q.popleft())
  #move  back to queue
  while stack:
    q.append(stack.pop())
#driver
q=deque([10,20,30])
print("original:",list(q))
reverse_queue(q)
print("reversed:",list(q))
print("reversed:",list(q))
```
